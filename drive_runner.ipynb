# 3. Start Background Services (Redis & Celery)

```
import os
import subprocess
import time

!sudo service redis-server start
time.sleep(2)

os.environ['CELERY_BROKER_URL'] = 'redis://localhost:6379/0'
os.environ['CELERY_RESULT_BACKEND'] = 'redis://localhost:6379/0'
subprocess.Popen('celery -A celery_worker.celery_app worker --loglevel=info', shell=True)
print('Redis and Celery started successfully!')
```

# 4. Start the Deepfake API (FastAPI + Ngrok)

```
import nest_asyncio
from pyngrok import ngrok
import uvicorn
from app import app
import getpass

print('Ngrok requires an authtoken to start a session.')
print('Get it here: https://dashboard.ngrok.com/get-started/your-authtoken')
authtoken = getpass.getpass('Enter your Ngrok Authtoken: ')
ngrok.set_auth_token(authtoken)

# Start an SSH tunnel on port 8000
tunnel = ngrok.connect(8000)
print(f'\n✅ COPY THIS URL INTO YOUR FRONTEND: {tunnel.public_url}')

# Run the FastAPI app
nest_asyncio.apply()
uvicorn.run(app, host="0.0.0.0", port=8000)
```
